# 01 NBA Data Exploration

Goal: learn what the NBA dataset contains before training any model.

In machine learning, this step is called exploratory data analysis, or EDA. EDA means we inspect the data carefully so we understand what each row represents, what columns are available, and whether the data is clean enough to model.

## Learning Checkpoint

Before writing model code, answer these questions:

1. What does one row represent?
2. Which columns identify the home and away teams?
3. Which columns contain the final scores?
4. Which column can become our label, `home_team_win`?

The label is the value we want the model to learn to predict.

## First Inspection Commands

These commands answer different questions:

- `head()` shows the first rows.
- `shape` shows row count and column count.
- `columns` lists the available fields.
- `info()` shows data types and missing values.
- `describe()` summarizes numeric columns.

Beginner mistake to avoid: do not train a model before you know what each important column means.
